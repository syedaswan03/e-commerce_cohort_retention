```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import matplotlib.image as mpimg
```

```python
cohort_retention_matrix = pd.read_csv('cohort_retention_matrix.csv', index_col=0)
cohort_retention_percentage = pd.read_csv('cohort_retention_percentage.csv', index_col=0)
plan_Segment = pd.read_csv('plan_segments.csv')
device_Segment = pd.read_csv('device_segments.csv')
country_Segment = pd.read_csv('country_segments.csv')
channel_Segment = pd.read_csv('channel_segments.csv')
```

```python
plt.figure(figsize=(12,8))

cohort_retention_percentage.head()
sns.heatmap(
    cohort_retention_percentage,
    annot=True,
    fmt=".1f",
    cmap="Blues"
)

plt.title("Customer Cohort Retention Heatmap")
plt.xlabel("Months Since First Purchase")
plt.ylabel("Cohort Month")
plt.tight_layout()
plt.show()
```

```python
average_retention = cohort_retention_matrix.mean(axis=0)

average_retention = average_retention.reset_index()
average_retention.columns = [
    "Month",
    "Retention"
]
```

```python
plt.figure(figsize=(10,6))
plt.plot(
    average_retention["Month"],
    average_retention["Retention"],
    marker="o",
    linewidth=3
)
plt.title("Average Customer Retention Decay Curve")
plt.xlabel("Months Since First Purchase")
plt.ylabel("Retention (%)")
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10,6))
plt.bar(channel_Segment["Acquisition_channel"], channel_Segment["CLTV"], color='skyblue')
plt.title("CLTV by Channel")
plt.xlabel("Acquisition Channel")
plt.ylabel("CLTV")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10,6))
plt.barh(country_Segment["Country"], country_Segment["CLTV"], color='navy')
plt.title("CLTV by Country")
plt.xlabel("CLTV")
plt.ylabel("Country")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10,6))
plt.bar(channel_Segment["Acquisition_channel"], channel_Segment["AOV"], color='forestgreen')
plt.title("AOV by Acquisition Channel")
plt.xlabel("AOV")
plt.ylabel("Acquisition Channel")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10,6))
plt.bar(channel_Segment["Acquisition_channel"], channel_Segment["Purchase_frequency_Channel"], color='crimson')
plt.title("Purchase Frequency by Acquisition Channel")
plt.xlabel("Purchase Frequency")
plt.ylabel("Acquisition Channel")
plt.tight_layout()
plt.show()
```

```python
# Load the images
heatmap = mpimg.imread("retention_heatmap.png")
curve = mpimg.imread("retention_decay_curve.png")
channel = mpimg.imread("CLTV_by_channel.png")
country = mpimg.imread("CLTV_by_country.png")
aov = mpimg.imread("AOV_by_channel.png")
frequency = mpimg.imread("Purchase_frequency_by_channel.png")
```

```python
# Image shapes
print(heatmap.shape)
print(curve.shape)
print(channel.shape)
print(country.shape)
print(aov.shape)
print(frequency.shape)
```

```python
# Create a dashboard with the images
fig, axes = plt.subplots(3, 2, figsize=(18,20))

axes[0,0].imshow(heatmap)
axes[0,0].set_title("Retention Heatmap")
axes[0,0].axis("off")

axes[0,1].imshow(curve)
axes[0,1].set_title("Retention Decay Curve")
axes[0,1].axis("off")

axes[1,0].imshow(channel)
axes[1,0].set_title("CLTV by Acquisition Channel")
axes[1,0].axis("off")

axes[1,1].imshow(country)
axes[1,1].set_title("CLTV by Country")
axes[1,1].axis("off")

axes[2,0].imshow(aov)
axes[2,0].set_title("Average Order Value")
axes[2,0].axis("off")

axes[2,1].imshow(frequency)
axes[2,1].set_title("Purchase Frequency")
axes[2,1].axis("off")

fig.suptitle(
    "E-Commerce Cohort Retention & CLTV Dashboard",
    fontsize=22,
    fontweight="bold"
)

plt.show()
```
