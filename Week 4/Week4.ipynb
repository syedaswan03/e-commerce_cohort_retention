```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
cohort_retention_matrix = pd.read_csv('cohort_retention_matrix.csv', index_col=0)
cohort_retention_percentage = pd.read_csv('cohort_retention_percentage.csv', index_col=0)
plan_Segment = pd.read_csv('plan_segments.csv')
device_Segment = pd.read_csv('device_segments.csv')
country_Segment = pd.read_csv('country_segments.csv')
channel_Segment = pd.read_csv('channel_segments.csv')
```

```python
plt.figure(figsize=(12,8))

cohort_retention_percentage.head()
sns.heatmap(
    cohort_retention_percentage,
    annot=True,
    fmt=".1f",
    cmap="Blues"
)



plt.title("Customer Cohort Retention Heatmap")
plt.xlabel("Months Since First Purchase")
plt.ylabel("Cohort Month")
plt.tight_layout()
plt.show()

plt.savefig(
    "retention_heatmap.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```
